1. Read the CSV file

```python
import pandas as pd
df = pd.read_csv("/content/pandas_dataset.csv")
print(df.shape)
df.head()
```

2. Rename Columns

```python
df.columns = df.columns = df.columns.str.lower()
print(df.columns)
```

3. Dropping rows

```python
df = df.dropna(how='all')
df = df.reset_index(drop=True)
df.head()
```

```python
df.head() #no columns dropped
```

5. counting missing values

```python
df.isna().sum()
```

6. Handling Missing Values

```python
df = df.dropna(subset=['sales'])
df['profit'] = df['profit'].fillna(df['profit'].mean())
df['discount'] = df['discount'].fillna(df['discount'].median())
df
```

```python
df['customer_name'] = df['customer_name'].fillna('Unknown')
df.isna().sum()
```

```python
duplicate_values = df.duplicated(subset=["order_id"])

print("Number of duplicated values:", duplicate_values.sum())

df = df.drop_duplicates(subset=["order_id"])
duplicate_count = df.duplicated(subset=["order_id"]).sum()
df.reset_index(drop=True, inplace=True)

print("Number of rows removed:", duplicate_count)
```

```python
df[df['unit_price'] > 20000]
df[(df['unit_price']>10000)&(df['status'] == 'completed')][['customer_name','category','status']]
df['total_amount'] = df['quantity'] * df['unit_price'] - df['discount']
df['customer_type'] = 'Regular Buyer'
df.loc[df['quantity'] >= 3, 'customer_type'] = 'Bulk Buyer'
```
