# Cleaning Process

Removing duplicates


```python
df = df.drop_duplicates().reset_index(drop=True)

print(f"Duplicates remaining: {df.duplicated().sum()}")
print(f"Total rows after removing duplicates: {df.shape[0]}")
```

Fixing the data-types

```python
# converting the date features into pandas date-time
df["published_date"] = pd.to_datetime(df["published_date"])

# shares sold are always in whole number
df["traded_quantity"] = df["traded_quantity"].astype(int)
```

Transforming the data


```python
# sorting by the date
df = df.sort_values(by="published_date").reset_index(drop=True)
```

```python
# extracting the year, month, and day from the pandas date-time
df["year"] = df["published_date"].dt.year
df["month"] = df["published_date"].dt.month
df["day_name"] = df["published_date"].dt.day_name()
```

*open and high are very important for NEPSE analysis to note the profit and loss in overall*

```python
df["price_change"] = (df["close"] - df["open"]).round(2)
df["daily_range"] = (df["high"] - df["low"]).round(2)
```

*Mapping the status for readibility*

```python
status_map = {1: "Gain", -1: "Loss", 0: "No Change"}
df["status_label"] = df["status"].map(status_map)
```

```python
# rounding off the traded amount
df["traded_amount"] = df["traded_amount"].round(2)
```

*Another important feature in trading analysis is, we see the price difference between yesterday's closing price and today's opening price*

```python
# getting yesterday's price with shift method
df['prev_close'] = df['close'].shift(1)

# calculating opening gap
df['opening_gap'] = (df['open'] - df['prev_close']).round(2)
```
